# Experiment 3 Analysis: CPU vs Latency & Reliability


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import glob

csv_files = glob.glob("cpu_pressure_results_*CORES.csv")
df = pd.concat([pd.read_csv(f) for f in csv_files], ignore_index=True)

df["cpu_cores"] = df["cpu_cores"].astype(float)
df = df.sort_values("cpu_cores")
```

## Latency vs CPU Cores


```python
plt.figure(figsize=(10, 6))
sns.lineplot(
    data=df[df["status"] == "success"],
    x="cpu_cores",
    y="latency_sec",
    marker="o"
)

plt.title("Processing Latency vs CPU Allocation")
plt.xlabel("CPU Cores")
plt.ylabel("Latency (Seconds)")
plt.grid(True)
plt.show()
```

## Reliability Under CPU Constraints


```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x="cpu_cores", hue="status")

plt.title("Pipeline Reliability vs CPU Allocation")
plt.xlabel("CPU Cores")
plt.ylabel("Number of Runs")
plt.show()
```
